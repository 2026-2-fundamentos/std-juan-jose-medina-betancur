```python
import pandas as pd

# Cargar la tabla original
timesheet = pd.read_csv('../data/timesheet.csv')

# Calcular el promedio de horas por conductor y conservar todos los registros
timesheet_with_means = timesheet.copy()
timesheet_with_means['mean_hours-logged'] = timesheet_with_means.groupby('driverId')['hours-logged'].transform('mean')

timesheet_with_means
```

```python
# Crear la tabla con el promedio de horas de cada conductor
timesheet_with_means = timesheet.copy()
timesheet_with_means['mean_hours-logged'] = timesheet_with_means.groupby('driverId')['hours-logged'].transform('mean')

timesheet_with_means
```

```python
# Sumar horas y millas por conductor
sum_timesheet = timesheet.groupby('driverId', as_index=False)[['hours-logged', 'miles-logged']].sum()
sum_timesheet
```

```python
# Calcular mínimo y máximo de horas registradas por conductor
min_max_timesheet = timesheet.groupby('driverId', as_index=False).agg(min_hours_logged=('hours-logged', 'min'), max_hours_logged=('hours-logged', 'max'))
min_max_timesheet
```

```python
# Combinar los totales por conductor con su nombre
drivers = pd.read_csv('../data/drivers.csv')
summary = sum_timesheet.merge(drivers[['driverId', 'name']], on='driverId', how='left')
summary = summary[['driverId', 'hours-logged', 'miles-logged', 'name']]
summary
```

```python
# Guardar summary como CSV, sin incluir el índice
summary.to_csv('../submission/summary.csv', index=False)
```

```python
# Seleccionar los diez conductores con mayor total de millas
top10 = summary.nlargest(10, 'miles-logged').reset_index(drop=True)
top10
```

```python
import matplotlib.pyplot as plt

# Graficar los diez conductores con más millas
top10_plot = top10.sort_values('miles-logged', ascending=True)
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(top10_plot['name'], top10_plot['miles-logged'])
ax.set_xlabel('Millas registradas')
ax.set_ylabel('Conductor')
ax.set_title('Top 10 conductores por millas registradas')
fig.tight_layout()
fig.savefig('../submission/top10_drivers.png', dpi=150, bbox_inches='tight')
plt.show()
```
